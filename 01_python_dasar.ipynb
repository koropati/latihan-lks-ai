## Bagian 2 — Variabel dan tipe data

**Variabel** adalah nama untuk sebuah nilai. Tanda `=` berarti "simpan nilai di

| `str` | teks (*string*) | `"192.168.1.1"`, `'Switch'` |
| `bool` | benar/salah | `True`, `False` |

Aturan nama variabel: huruf kecil, pakai `_` sebagai pemisah kata, tanpa spasi,

## Bagian 3 — Operator

**Aritmetika:** `+` `-` `*` `/` `//` (bagi bulat) `%` (sisa bagi) `**` (pangkat)

**Pembanding:** `==` `!=` `<` `>` `<=` `>=` — semuanya menghasilkan `True`/`False`

**Logika:** `and` (kedua benar), `or` (salah satu benar), `not` (kebalikan)

Perhatikan: `=` menyimpan nilai, `==` membandingkan nilai. Tertukar keduanya
adalah galat pemula nomor satu.

````python
a, b = 17, 5

print("a + b  =", a + b)
print("a / b  =", a / b)     # 3.4  -> hasil / selalu float
print("a // b =", a // b)    # 3    -> dibulatkan ke bawah
print("a % b  =", a % b)     # 2    -> sisa bagi
print("a ** 2 =", a ** 2)    # 289

print()
print("a > b        :", a > b)
print("a == 17      :", a == 17)
print("a > b and b > 10 :", a > b and b > 10)
print("not (a > b)  :", not (a > b))
````

`%` (sisa bagi) lebih berguna daripada kelihatannya. Contoh khas: menentukan
ganjil/genap, atau memproses setiap elemen ke-n.

````python
for n in [10, 11, 12, 13]:
    jenis = "genap" if n % 2 == 0 else "ganjil"
    print(f"{n} itu {jenis}")
````

## Bagian 4 — Percabangan: `if`, `elif`, `else`

Program perlu mengambil keputusan berbeda untuk kondisi berbeda.

Dua hal yang wajib benar, kalau tidak Python menolak menjalankannya:

1. Ada **titik dua** `:` di akhir baris `if`.
2. Baris di bawahnya **menjorok masuk** (indentasi) — gunakan 4 spasi, konsisten.
   Di Colab, tombol Tab sudah menghasilkan 4 spasi.

Urutan pemeriksaan penting: Python berhenti pada cabang pertama yang benar.

````python
latensi = 85   # ms — ubah nilai ini lalu jalankan ulang selnya

if latensi < 30:
    status = "sangat baik"
elif latensi < 100:
    status = "wajar"
elif latensi < 300:
    status = "lambat"
else:
    status = "bermasalah"

print(f"Latensi {latensi} ms -> jaringan {status}")
````

Fungsi di bawah ini akan kita pakai lagi di Bagian 6. Perhatikan bahwa yang
dilakukannya sama dengan pekerjaan sebuah model **klasifikasi**: menerima
masukan angka, mengembalikan satu label dari beberapa pilihan.

Bedanya besar sekali, dan ini inti dari seluruh pelatihan ini:

- Di sini **kamu** yang menuliskan batas 30, 100, dan 300.
- Pada machine learning, **batas itu ditemukan oleh program** dari data berlabel.

Kalau batasnya jelas dan sedikit, aturan `if` seperti ini justru lebih baik:
lebih cepat, gratis, dan bisa diperiksa orang lain. Machine learning masuk
ketika aturannya terlalu banyak atau tidak diketahui — misalnya membedakan foto
sampah organik dan anorganik, yang tidak mungkin ditulis sebagai rangkaian `if`.

````python
def nilai_status(latensi_ms):
    if latensi_ms < 30:
        return "sangat baik"
    elif latensi_ms < 100:
        return "wajar"
    elif latensi_ms < 300:
        return "lambat"
    return "bermasalah"


for nilai in [12, 45, 150, 800]:
    print(f"{nilai:>4} ms -> {nilai_status(nilai)}")
````

## Bagian 5 — Perulangan: `for` dan `while`

`for` dipakai bila jumlah putarannya sudah diketahui — misalnya sepanjang sebuah
daftar. `while` dipakai bila putaran berhenti karena sebuah kondisi.

````python
# for atas sebuah daftar
perangkat = ["Router-Utama", "Switch-Lab-01", "AP-Perpustakaan"]

for p in perangkat:
    print("Memeriksa", p)

print()

# range(1, 5) menghasilkan 1, 2, 3, 4 — angka terakhir TIDAK ikut
for i in range(1, 5):
    print(f"192.168.10.{i}")

print()

# enumerate memberi nomor urut sekaligus isinya
for nomor, p in enumerate(perangkat, start=1):
    print(f"{nomor}. {p}")
````

````python
# while: berhenti ketika kondisinya tidak lagi benar
percobaan = 0
maksimal = 3
berhasil = False

while percobaan < maksimal and not berhasil:
    percobaan += 1                      # += artinya "tambahkan ke dirinya sendiri"
    print(f"Percobaan koneksi ke-{percobaan}...")
    if percobaan == 2:                  # anggap percobaan kedua berhasil
        berhasil = True

print("Berhasil" if berhasil else "Gagal setelah 3 percobaan")
````

> **Hati-hati dengan `while`.** Kalau kondisinya tidak pernah menjadi salah,
> selnya berjalan selamanya. Di Colab, hentikan dengan tombol **■ (stop)** di
> sebelah kiri sel.

Dua kata kunci pengendali perulangan:

- `break` — keluar dari perulangan sekarang
- `continue` — lompat ke putaran berikutnya

````python
log = [12, 45, -1, 88, 950, 33]   # -1 menandakan tidak ada balasan (timeout)

for nilai in log:
    if nilai == -1:
        continue                   # lewati data rusak, jangan diikutkan
    if nilai > 900:
        print(f"Latensi {nilai} ms terlalu tinggi — pemeriksaan dihentikan.")
        break
    print(f"{nilai} ms oke")
````

`range`.

## Bagian 6 — `list`, `dict`, `tuple`, `set`

Empat wadah untuk menyimpan banyak nilai. Yang paling sering kamu pakai: `list`
dan `dict`.

| Wadah | Bentuk | Dipakai untuk |
|---|---|---|
| `list` | `[1, 2, 3]` | urutan yang bisa diubah |
| `dict` | `{"nama": "nilai"}` | pasangan kunci → nilai |
| `tuple` | `(1, 2)` | urutan yang tidak boleh diubah |
| `set` | `{1, 2, 3}` | kumpulan nilai unik, tanpa urutan |

````python
# ---------- LIST ----------
ip = ["192.168.1.1", "192.168.1.10", "192.168.1.20"]

print("Isi pertama :", ip[0])      # penomoran dimulai dari 0
print("Isi terakhir:", ip[-1])     # indeks negatif = dihitung dari belakang
print("Dua pertama :", ip[0:2])    # potongan (slice): indeks 0 dan 1
print("Jumlah isi  :", len(ip))

ip.append("192.168.1.30")          # tambah di akhir
print("Setelah append:", ip)

print("Apakah .1.10 ada?", "192.168.1.10" in ip)
````

````python
# ---------- DICT ----------
switch = {
    "nama": "Switch-Lab-01",
    "ip": "192.168.1.10",
    "port": 24,
    "port_terpakai": 18,
}

print(switch["nama"])                       # ambil nilai lewat kuncinya
print("Sisa port:", switch["port"] - switch["port_terpakai"])

switch["lokasi"] = "Lab Komputer 1"         # menambah kunci baru
print(switch)

print()
# menelusuri dict: .items() memberi kunci dan nilai sekaligus
for kunci, nilai in switch.items():
    print(f"{kunci:<15}: {nilai}")
````

`.get()` lebih aman daripada `[...]`: kalau kuncinya tidak ada, `[...]` memicu
galat `KeyError`, sedangkan `.get()` mengembalikan nilai cadangan. Ini penting
saat mengolah data nyata yang sering tidak lengkap.

````python
print(switch.get("vlan"))                 # None — kunci "vlan" tidak ada
print(switch.get("vlan", "belum diatur")) # nilai cadangan
````

````python
# ---------- TUPLE dan SET ----------
koordinat = (-8.17, 115.09)          # tuple: tidak bisa diubah setelah dibuat
print("Lintang:", koordinat[0])

jenis_perangkat = ["switch", "router", "switch", "ap", "router", "switch"]
unik = set(jenis_perangkat)          # set membuang duplikat
print("Jenis unik:", unik)
print("Ada berapa jenis:", len(unik))
````

### Gabungan `list` + `dict`: bentuk data yang paling sering kamu temui

Satu baris data = satu `dict`. Banyak baris = `list` berisi `dict`. Inilah
bentuk yang nanti diubah menjadi tabel `DataFrame` di Notebook 02.

````python
data_jaringan = [
    {"ruang": "Lab Komputer 1", "latensi_ms": 12,  "paket_hilang": 0.0},
    {"ruang": "Lab Komputer 2", "latensi_ms": 45,  "paket_hilang": 0.4},
    {"ruang": "Perpustakaan",   "latensi_ms": 150, "paket_hilang": 2.1},
    {"ruang": "Aula",           "latensi_ms": 320, "paket_hilang": 7.5},
]

print(f"{'ruang':<18}{'latensi':>9}{'hilang %':>10}  status")
print("-" * 55)
for baris in data_jaringan:
    print(f"{baris['ruang']:<18}{baris['latensi_ms']:>9}{baris['paket_hilang']:>10.1f}"
          f"  {nilai_status(baris['latensi_ms'])}")

# rata-rata latensi
semua_latensi = [b["latensi_ms"] for b in data_jaringan]   # list comprehension
print()
print("Rata-rata latensi:", sum(semua_latensi) / len(semua_latensi), "ms")
````

Baris `[b["latensi_ms"] for b in data_jaringan]` disebut **list comprehension** —
cara singkat menulis perulangan yang menghasilkan list baru. Setara dengan:

```python
semua_latensi = []
for b in data_jaringan:
    semua_latensi.append(b["latensi_ms"])
```

Bentuk singkat ini sering muncul di kode machine learning, jadi kenali saja
bentuknya. Kalau membingungkan, tulis versi panjangnya — hasilnya sama.

## Bagian 7 — Fungsi

Fungsi adalah kode yang dibungkus dan diberi nama supaya bisa dipakai berulang.
Alasan memakainya: kalau logikanya perlu diperbaiki, cukup diperbaiki di satu
tempat.

- `def nama(parameter):` — mendefinisikan
- `return` — mengembalikan hasil ke pemanggil
- parameter boleh punya nilai bawaan (*default*)

````python
def ringkas(nilai, satuan="ms"):
    'Mengembalikan rata-rata, nilai terkecil, dan terbesar dari sebuah list.'
    if len(nilai) == 0:
        return None                    # lindungi dari pembagian dengan nol
    return {
        "n": len(nilai),
        "rata_rata": sum(nilai) / len(nilai),
        "min": min(nilai),
        "max": max(nilai),
        "satuan": satuan,
    }


hasil = ringkas(semua_latensi)
print(hasil)
print()
print(f"Dari {hasil['n']} ruang: rata-rata {hasil['rata_rata']:.1f} {hasil['satuan']}, "
      f"terendah {hasil['min']}, tertinggi {hasil['max']}")

print()
print("List kosong ->", ringkas([]))   # tidak galat, karena sudah dijaga
````

Teks dalam tanda kutip di baris pertama fungsi disebut **docstring** —
penjelasan singkat tentang apa yang dilakukan fungsi itu. Tulislah; dirimu di
masa depan akan berterima kasih.

## Bagian 8 — Membaca berkas CSV

CSV (*comma-separated values*) adalah berkas teks biasa: baris pertama nama
kolom, baris berikutnya datanya, dipisah koma. Hampir semua data terbuka
tersedia dalam bentuk ini.

Sel di bawah membuat berkas CSV kecil supaya notebook ini bisa dijalankan tanpa
mengunggah apa pun. Di Notebook 02 kita akan membaca CSV sungguhan dari internet.

````python
isi_csv = '''tanggal,ruang,jenis_ruang,jam_pakai,kwh
2026-08-03,Lab Komputer 1,lab,7.7,13.3
2026-08-03,Lab Komputer 2,lab,2.3,3.72
2026-08-03,Kelas X TKJ 1,kelas,6.0,4.10
2026-08-04,Lab Komputer 1,lab,8.1,14.05
2026-08-04,Kelas X TKJ 1,kelas,5.5,3.80
2026-08-04,Perpustakaan,umum,4.0,2.95
'''

with open("contoh_listrik.csv", "w", encoding="utf-8") as f:
    f.write(isi_csv)

print("Berkas contoh_listrik.csv dibuat.")
````

`with open(...) as f:` adalah cara baku membuka berkas. Keunggulannya: berkas
otomatis ditutup ketika blok `with` selesai, walaupun terjadi galat di tengah
jalan.

Modul `csv` adalah bagian dari Python sendiri — tidak perlu dipasang.
`csv.DictReader` membaca setiap baris menjadi `dict`, dengan nama kolom sebagai
kuncinya. Bentuknya persis seperti `data_jaringan` di Bagian 6.

````python
import csv

with open("contoh_listrik.csv", encoding="utf-8") as f:
    baris = list(csv.DictReader(f))

print("Jumlah baris:", len(baris))
print("Nama kolom  :", list(baris[0].keys()))
print()
print("Baris pertama:", baris[0])
````

Ingat Bagian 2: **semua nilai dari CSV terbaca sebagai teks.** Kolom `kwh` di
atas berisi `'13.3'`, bukan `13.3`. Harus dikonversi sebelum dihitung.

````python
print("Tipe nilai kwh dari CSV:", type(baris[0]["kwh"]))   # <class 'str'>

# Hitung total dan rata-rata kWh per jenis ruang
total, jumlah = {}, {}
for b in baris:
    jenis = b["jenis_ruang"]
    total[jenis] = total.get(jenis, 0.0) + float(b["kwh"])   # float() wajib di sini
    jumlah[jenis] = jumlah.get(jenis, 0) + 1

print()
print(f"{'jenis':<8}{'total kWh':>11}{'rata-rata':>11}{'baris':>7}")
print("-" * 37)
for jenis in sorted(total, key=total.get, reverse=True):
    print(f"{jenis:<8}{total[jenis]:>11.2f}{total[jenis]/jumlah[jenis]:>11.2f}{jumlah[jenis]:>7}")
````

## Bagian 9 — Galat dan cara membacanya

Galat itu normal, bukan tanda kamu tidak bisa. Yang membedakan pemula dan
bukan pemula hanyalah kecepatan membaca pesannya.

**Baca pesan galat dari baris paling bawah.** Baris terakhir menyebutkan jenis
galat dan alasannya. Baris di atasnya menunjukkan di mana galat itu terjadi.

Lima galat yang paling sering muncul:

| Pesan | Artinya | Perbaikannya |
|---|---|---|
| `NameError` | nama variabel belum ada | jalankan sel di atasnya, atau cek ejaan |
| `TypeError` | tipe data tidak cocok | konversi dengan `int()` / `float()` / `str()` |
| `IndexError` | indeks list melebihi isinya | cek `len()` dulu |
| `KeyError` | kunci dict tidak ada | pakai `.get()`, atau cek `.keys()` |
| `IndentationError` | indentasi tidak konsisten | pakai 4 spasi, jangan campur dengan Tab |
| `FileNotFoundError` | berkas tidak ada di folder itu | cek nama dan lokasinya |

`try` / `except` dipakai kalau kamu **sudah menduga** sebuah galat bisa terjadi
dan tahu apa yang harus dilakukan sebagai gantinya. Contoh khas: data kotor dari
berkas.

Jangan pakai `try/except` untuk menyembunyikan galat yang belum kamu pahami —
itu memindahkan masalahnya ke tempat yang lebih sulit ditemukan.

````python
data_kotor = ["13.3", "3.72", "", "4.10", "tidak terbaca", "2.95"]

bersih, ditolak = [], []
for nilai in data_kotor:
    try:
        bersih.append(float(nilai))
    except ValueError:
        ditolak.append(repr(nilai))        # repr() supaya string kosong kelihatan

print("Terbaca :", bersih)
print("Ditolak :", ditolak)
print(f"{len(bersih)} dari {len(data_kotor)} nilai berhasil dibaca.")
````

Cara melaporkannya patut diperhatikan: **"4 dari 6 nilai berhasil dibaca"** jauh
lebih berguna daripada "data sudah dibersihkan". Kebiasaan menyebut jumlah ini
akan dipakai terus sampai Notebook 04, dan juga saat tanya jawab dengan dewan
juri.
